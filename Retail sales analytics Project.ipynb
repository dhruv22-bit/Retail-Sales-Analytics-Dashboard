```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
import numpy as np
```

```python
orders = pd.read_csv(r"C:\Users\chanc\Downloads\Orders_Big.csv")
products = pd.read_csv(r"C:\Users\chanc\Downloads\Products_Big.csv")
customers = pd.read_csv(r"C:\Users\chanc\Downloads\Customers_Big.csv")
```

```python
df = orders.merge(products,on="ProductID")
df = df.merge(customers,on="CustomerID")
```

```python
df["Revenue"] = df["Quantity"] * df["SellingPrice"]
```

```python
df["Profit"] = (df["SellingPrice"] -df["CostPrice"]) * df["Quantity"]
```

Category with maximum sales.

```python
df.groupby("Category")["Revenue"].sum()
```

Most profitable region

```python
df.groupby("Region")["Profit"].sum()
```

Most amount spending customers

```python
df.groupby("CustomerName")["Revenue"].sum()
```

```python
sns.barplot(data=df, x="Discount", y="Revenue")
plt.title("Revenue vs Discount")
```

Does delivery time affect ratings

```python
sns.set_style("whitegrid")
sns.boxplot(data = df , x = "DeliveryDays", y = "Rating")
plt.title("Rating vs DeliveryDays")
plt.show()
```
